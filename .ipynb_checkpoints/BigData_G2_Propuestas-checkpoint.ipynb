## Propuesta 1: DataCo Smart Supply Chain for Big Data Analysis

- **Enlace:** [Kaggle - DataCo Smart Supply Chain](https://www.kaggle.com/datasets/shashwatwork/dataco-smart-supply-chain-for-big-data-analysis)
- **Tamaño:** ~180,000 filas y 96 columnas
- **Dominio:** Cadena de suministro / logística
- **Justificación:** Dataset alineado con sistemas reales de nivel empresarial (Salesforce, SAP), con los que uno de los integrantes ha trabajado directamente. Su tamaño y variedad de columnas lo hacen adecuado para un análisis de Big Data robusto.

## Propuesta 2: Customer Support Ticket Dataset

- **Enlace:** [Kaggle - Customer Support Ticket Dataset](https://www.kaggle.com/datasets/suraj520/customer-support-ticket-dataset)
- **Tamaño:** ~29,000 filas y 17 columnas
- **Dominio:** Atención al cliente / soporte técnico
- **Justificación:** Todos los integrantes del grupo están relativamente familiarizados con este tipo de datos (tickets de soporte), lo que facilita el análisis y la interpretación de resultados.

## Propuesta 3: CRM Sales Opportunities

- **Enlace:** [Kaggle - CRM Sales Opportunities](https://www.kaggle.com/datasets/innocentmfa/crm-sales-opportunities)
- **Tamaño:** *~86 filas y 7 columnas
- **Dominio:** Ventas / CRM
- **Justificación:** Uno de los integrantes trabaja actualmente en un proceso similar en su empresa, lo que da contexto real al análisis y permite que el proyecto sirva como pieza de portafolio profesional.
